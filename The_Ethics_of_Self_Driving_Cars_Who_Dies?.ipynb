```python
country=pd.read_csv("country_preferences.csv")
```

```python
print(country.head())
```

```python
print(country.columns.tolist())
```

```python
cluster_lookup = country[["country", "cultural_cluster"]].drop_duplicates()
print(cluster_lookup.shape)
print(cluster_lookup.head())
```

```python
print(cluster_lookup["country"].value_counts().head(10))
```

```python
resp_countries = set(responses["country"].dropna().unique())
agg_countries = set(cluster_lookup["country"].unique())
print(agg_countries - resp_countries)
```

```python
#how="left" means:
#"keep every row from responses no matter what —
#if there's no matching country in cluster_lookup,
#just fill cultural_cluster with NaN instead of dropping the row."
responses_with_cluster = pd.merge(
    responses,
    cluster_lookup,
    on="country",
    how="left"
)
print(responses_with_cluster.shape)
print(responses_with_cluster[["country", "cultural_cluster"]].head())
```

```python
responses_with_cluster.shape
```

```python
responses_with_cluster[responses_with_cluster["scenario_type"] == "Age"]
```

```python
filtered = responses_with_cluster[
    (responses_with_cluster["scenario_type"] == "Age") &
    (responses_with_cluster["character_group"] == "Young")
]
```

```python
filtered.groupby("cultural_cluster")["saved"].mean()
```
